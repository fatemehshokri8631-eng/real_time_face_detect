```python
loop=True

while loop:
 
 _, img_webcam=cam.read()
 gray=cv2.cvtColor(img_webcam,cv2.COLOR_BGR2GRAY)
 face=face_model.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)


 if len(face) > 0:
     x=face[0][0]
     y=face[0][1]
     x2=x+ face[0][2]
     y2=y+ face[0][3]



 gray_face=gray[y:y2,x:x2]
 eye=eye_model.detectMultiScale(gray_face,minSize=(30,30),scaleFactor=1.1,minNeighbors=5)
 imgout=img_webcam.copy()

 out=cv2.rectangle(imgout, (x,y), (x2,y2), (0,250,0),3)

 ic=0
 for (xe, ye, w ,h) in eye:
    ic=ic+1
    cv2.rectangle(imgout, (xe+x, ye+y), (xe+w+y, ye+h+y), (0 ,0 ,250), 2)
    if ic==2:
       break


 cv2.imshow('faty', out)

 if cv2.waitKey(1) == ord('q'):
     loop=False
     cv2.destroyAllWindows()
     break






#  if len(eye) >= 2:
#      xb=eye[6][0]
#      yb=eye[6][1]
#      xb2=x+ eye[6][2]
#      yb2=y+ eye[6][3]

#      xc=eye[1][0]
#      yc=eye[1][1]
#      xc2=x+ eye[1][2]
#      yc2=y+ eye[1][3]

#      out=cv2.rectangle(img_webcam,(xb,yb),(xb2,yb2),(0,250,0),3)
#      out=cv2.rectangle(img_webcam,(xc,yc),(xc2,yc2),(250,0,0),3)

```
